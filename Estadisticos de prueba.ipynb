```python
import numpy as np
import scipy.stats as stats
import pandas as pd
from statsmodels.stats.anova import anova_lm
from statsmodels.formula.api import ols
```

```python
# Pregunta 1: ¿Tienen los estudiantes de 7mo cuatrimestre una estatura media diferente de 175 cm?

media_muestral = 171.94
media_esperada = 175
desviacion_estandar = 3.91
n = 30  # Tamaño de muestra

# Muestra aleatoria basada en los datos proporcionados
muestra = np.random.normal(media_muestral, desviacion_estandar, n)

# Prueba t de una muestra
t_stat, p_value = stats.ttest_1samp(muestra, media_esperada)

print("Hipótesis 1: ¿Tienen los estudiantes de 7mo cuatrimestre una estatura media diferente de 175 cm?")
print(f"t-statistic: {t_stat}, p-value: {p_value}")
if p_value < 0.05:
    print("Se rechaza la hipótesis nula: la estatura media es significativamente diferente de 175 cm.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia significativa para decir que la media es diferente de 175 cm.\n")
```

```python
# Pregunta 2: ¿Es la desviación estándar de su estatura igual o menor que 10 cm?

# Datos de la Hipótesis 2
var_muestral = desviacion_estandar ** 2
var_esperada = 10 ** 2
chi2_stat = (n - 1) * var_muestral / var_esperada

# Prueba Chi-cuadrada para la varianza
p_value_chi2 = stats.chi2.sf(chi2_stat, df=n-1)

print("Hipótesis 2: ¿Es la desviación estándar de su estatura igual o menor que 10 cm?")
print(f"Chi-cuadrado: {chi2_stat}, p-value: {p_value_chi2}")
if p_value_chi2 < 0.05:
    print("Se rechaza la hipótesis nula: la desviación estándar es significativamente menor que 10 cm.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia suficiente para decir que la desviación estándar es menor que 10 cm.\n")
```

```python
# Pregunta 3: ¿Es diferente la estatura promedio de las mujeres comparada con la de los hombres?

media_hombres = 171.93
media_mujeres = 169
desviacion_estandar_comun = 3.91
n_hombres = n_mujeres = 30  # Tamaño de muestra para los dos grupos

muestra_hombres = np.random.normal(media_hombres, desviacion_estandar_comun, n_hombres)
muestra_mujeres = np.random.normal(media_mujeres, desviacion_estandar_comun, n_mujeres)

# Prueba t de dos muestras independientes
t_stat_2, p_value_2 = stats.ttest_ind(muestra_hombres, muestra_mujeres)

print("Hipótesis 3: ¿Es diferente la estatura promedio de las mujeres comparada con la de los hombres?")
print(f"t-statistic: {t_stat_2}, p-value: {p_value_2}")
if p_value_2 < 0.05:
    print("Se rechaza la hipótesis nula: la estatura promedio entre hombres y mujeres es significativamente diferente.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia suficiente para decir que la estatura promedio es diferente entre hombres y mujeres.\n")
```

```python
# Pregunta 4: ¿Hay diferencias significativas en los retrasos de llegada entre la Aerolínea A y la Aerolínea B?

# Datos de la Hipótesis 4
media_A = 15
media_B = 25
desviacion_A = 10
desviacion_B = 15
n_A = n_B = 100  # Tamaño de muestra para las dos aerolíneas

muestra_A = np.random.normal(media_A, desviacion_A, n_A)
muestra_B = np.random.normal(media_B, desviacion_B, n_B)

# Prueba t de dos muestras independientes
t_stat_3, p_value_3 = stats.ttest_ind(muestra_A, muestra_B)

print("Hipótesis 4: ¿Hay diferencias significativas en los retrasos de llegada entre la Aerolínea A y la Aerolínea B?")
print(f"t-statistic: {t_stat_3}, p-value: {p_value_3}")
if p_value_3 < 0.05:
    print("Se rechaza la hipótesis nula: hay diferencias significativas en los retrasos de llegada entre Aerolínea A y Aerolínea B.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia suficiente para decir que los retrasos de llegada son diferentes entre las aerolíneas.\n")
```

```python
# Pregunta 5: ¿El retraso promedio de llegada varía según el día de la semana?

data = pd.DataFrame({
    'dia': ['Lunes']*200 + ['Martes']*200 + ['Miércoles']*200 + ['Jueves']*200 + ['Viernes']*200 + ['Sábado']*200 + ['Domingo']*200,
    'retraso': np.random.normal(20, 15, 200).tolist() + np.random.normal(15, 10, 200).tolist() +
               np.random.normal(25, 20, 200).tolist() + np.random.normal(10, 5, 200).tolist() +
               np.random.normal(30, 25, 200).tolist() + np.random.normal(5, 2, 200).tolist() +
               np.random.normal(12, 7, 200).tolist()
})

# ANOVA de una vía
modelo = ols('retraso ~ dia', data=data).fit()
anova_resultado = anova_lm(modelo)

print("Hipótesis 5: ¿El retraso promedio de llegada varía según el día de la semana?")
print(anova_resultado)

if anova_resultado["PR(>F)"].iloc[0] < 0.05:
    print("Se rechaza la hipótesis nula: el retraso promedio varía según el día de la semana.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia suficiente para decir que el retraso varía según el día.\n")
```

```python
# Pregunta 6: ¿Existe una correlación significativa entre el retraso de salida y el retraso de llegada?

retraso_salida = np.random.normal(20, 5, 150)
retraso_llegada = np.random.normal(25, 7, 150)

# Prueba de correlación de Pearson
coef_corr, p_value_corr = stats.pearsonr(retraso_salida, retraso_llegada)

print("Hipótesis 6: ¿Existe una correlación significativa entre el retraso de salida y el retraso de llegada?")
print(f"Coeficiente de correlación: {coef_corr}, p-value: {p_value_corr}")
if p_value_corr < 0.05:
    print("Se rechaza la hipótesis nula: existe una correlación significativa entre el retraso de salida y el retraso de llegada.\n")
else:
    print("No se rechaza la hipótesis nula: no hay evidencia suficiente para decir que existe una correlación significativa.\n")
```
